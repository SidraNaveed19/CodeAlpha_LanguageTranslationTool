```python
import gradio as gr
from deep_translator import GoogleTranslator

LANGUAGES = {
    "English": "en",
    "Spanish": "es",
    "French": "fr",
    "German": "de",
    "Arabic": "ar",
    "Hindi": "hi",
    "Chinese (Simplified)": "zh-CN"
}

def translate_text(text, source_lang, target_lang):
    if not text or not text.strip():
        return "Please enter text to translate."
    try:
        src_code = LANGUAGES[source_lang]
        dest_code = LANGUAGES[target_lang]
        translated = GoogleTranslator(source=src_code, target=dest_code).translate(text)
        return translated
    except Exception as e:
        return f"Translation Error: {str(e)}"

with gr.Blocks() as demo:
    gr.Markdown("# CodeAlpha AI - Language Translation Tool")

    with gr.Row():
        src_menu = gr.Dropdown(choices=list(LANGUAGES.keys()), value="English", label="Source Language")
        dest_menu = gr.Dropdown(choices=list(LANGUAGES.keys()), value="Spanish", label="Target Language")

    with gr.Row():
        input_text = gr.Textbox(lines=5, placeholder="Enter text to translate...", label="Input Text")
        output_text = gr.Textbox(lines=5, interactive=False, label="Translated Result")

    submit_btn = gr.Button("Translate Text", variant="primary")
    submit_btn.click(fn=translate_text, inputs=[input_text, src_menu, dest_menu], outputs=output_text)

demo.launch(share=True)
```
